# Phase 3 core modeling
This reporting-only notebook consumes compact saved JSON evidence and never trains a model. Validation-v1 remained frozen after the Phase 2 LightGBM reference (OOF 0.592987) and its first public score exposed a public–local generalization gap.

P3-D1 tested only recent observed horizon-1 rows, so it is secondary evidence. Drift showed raw year outside the sampled training range for 100% of recent and Test rows, while measured hydrological/spatial predictors were mostly in range.

```python
import json
from pathlib import Path


def metric(path): return json.loads(Path(path).read_text())
m1=metric('../reports/phase3m1_metrics.json'); m2=metric('../reports/phase3m2_metrics.json')
m3=metric('../reports/phase3m3b_metrics.json'); m4=metric('../reports/phase3m4_metrics.json')
e1=metric('../reports/phase3e1_metrics.json'); prod=metric('../reports/phase3h_production_manifest.json')
[(x['experiment_id'], x['metrics']['pooled']['rmse']) for x in (m1,m2,m3,m4,e1)]
```

## Evidence and decisions
Absolute prediction was replaced by change-from-last-observed residual prediction in P3-M1. P3-M2 added temporal, seasonal, soil-anomaly, contrast, and spherical features but weakened protected long horizons. P3-M3B removed only raw year from P3-M1 and passed every gate (0.583923). Specialists improved horizon 1 but failed horizon 5 protection. The cross-fitted six-component ensemble had the best pooled score (0.571686) but failed horizon 6 protection and was rejected.

The production selection is therefore unchanged P3-M3B. The primary and persistence diagnostic candidates were independently validated but not uploaded. Limitations include a two-fold validation design, Test year extrapolation, P3-D1's restricted population, sparse extreme-condition bins, and no hidden-target evidence. Seasonal anomaly and post-result hybrid repairs were rejected.
